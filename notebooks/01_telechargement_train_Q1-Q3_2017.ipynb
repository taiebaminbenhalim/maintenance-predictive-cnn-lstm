```python
import requests
import zipfile
import pandas as pd
import io
import os
import tempfile
```

```python
def process_backblaze_data(zip_url):
    """
    Downloads a zip file from Backblaze, extracts CSVs, and concatenates them into a single DataFrame.
    """
    print(f"Downloading {zip_url}...")
    response = requests.get(zip_url)
    response.raise_for_status() # Raise an exception for HTTP errors

    # Create a temporary directory to extract files
    with tempfile.TemporaryDirectory() as temp_dir:
        print(f"Extracting files to temporary directory: {temp_dir}")

        # Read the content into a BytesIO object and extract
        with zipfile.ZipFile(io.BytesIO(response.content)) as zf:
            zf.extractall(temp_dir)

        print("Extraction complete. Identifying and concatenating CSV files...")

        all_csv_files = []
        for root, _, files in os.walk(temp_dir):
            for file in files:
                file_path = os.path.join(root, file)
                # Exclude files from __MACOSX directory and files starting with ._
                if file_path.endswith('.csv') and '__MACOSX' not in file_path and not file.startswith('._'):
                    all_csv_files.append(file_path)

        if not all_csv_files:
            print("No CSV files found in the extracted content.")
            return pd.DataFrame() # Return empty DataFrame if no CSVs found
        else:
            # Read and concatenate all CSV files
            df_list = []
            for csv_file in all_csv_files:
                print(f"Reading {csv_file}...")
                df_list.append(pd.read_csv(csv_file))

            combined_df = pd.concat(df_list, ignore_index=True)
            print("All CSV files concatenated successfully.")
            return combined_df
```

```python
# URL for Q1 2017 data
zip_url_q1 = "https://f001.backblazeb2.com/file/Backblaze-Hard-Drive-Data/data_Q1_2017.zip"

# Process the Q1 data using the new function
print("\n--- Processing Q1 2017 Data ---")
combined_df_q1 = process_backblaze_data(zip_url_q1)
```

```python
# URL for Q2 2017 data
zip_url_q2 = "https://f001.backblazeb2.com/file/Backblaze-Hard-Drive-Data/data_Q2_2017.zip"

# Process the Q2 data using the new function
print("\n--- Processing Q2 2017 Data ---")
combined_df_q2 = process_backblaze_data(zip_url_q2)
```

```python
# URL for Q3 2017 data
zip_url_q3 = "https://f001.backblazeb2.com/file/Backblaze-Hard-Drive-Data/data_Q3_2017.zip"

# Process the Q3 data using the function
print("\n--- Processing Q3 2017 Data ---")
combined_df_q3 = process_backblaze_data(zip_url_q3)
```
